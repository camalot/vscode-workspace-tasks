This notebook demonstrates how different languages can be executed within a Jupyter environment. It includes simple examples for Python, JavaScript, and Bash.

## Python Example

This cell executes a simple Python script to calculate the first 10 numbers of the Fibonacci sequence.

```python
def fibonacci(n):
    sequence = [0, 1]
    while len(sequence) < n:
        next_val = sequence[-1] + sequence[-2]
        sequence.append(next_val)
    return sequence

print(f"First 10 Fibonacci numbers: {fibonacci(10)}")
```

## JavaScript Example

This cell runs JavaScript code (requires a Node.js or compatible kernel) to perform a basic array mapping.

```python
const numbers = [1, 2, 3, 4, 5];
const squared = numbers.map(num => num * num);

console.log('Original:', numbers);
console.log('Squared:', squared);
```
